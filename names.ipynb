Let's create a name generator. Each name comprises of `n` words from the mnemonic wordlist above. Users choose `n`, a seed, a starting position and the number of successive names they want. The generator must support random access to HUGE namespaces and guarantee no duplicate names.

Here's the approach I've got in mind:

1. Calculate the namespace size with \(P(1633,n)\).
2. Treat each possible name as an integer rank.
3. Use a seeded reversible Feistel permutation, with cycle walking to cover exactly the valid rank range.
4. Convert each permuted rank to `n` distinct word indices using factoradic unranking.
5. Map those indices to words.

This should provide deterministic, duplicate-free, stateless, random access.

Can you restate the understanding of the problem in your own words. How does the approach sound? Anything missing or any gotchas?

##### 🤖Reply🤖<!-- SOLVEIT_SEPARATOR_7f3a9b2c -->

You want a deterministic function

\[
(n,\ \text{seed},\ \text{position}) \longmapsto \text{name}
\]

where each name contains exactly `n` **distinct** words from the 1,633-word list. For a fixed `n` and seed, positions should behave like a shuffled enumeration of every valid ordered name:

- position `0` gives one name,
- position `1` gives another,
- random access works without generating earlier names,
- no valid name is repeated,
- the result is reproducible from the same inputs,
- and the generator needs no mutable state.

The number of valid ordered names is the falling factorial

\[
P(1633,n)=1633\cdot1632\cdots(1633-n+1).
\]

You then intend to:

1. interpret the position as a rank in `[0, P(1633,n))`;
2. apply a seed-dependent permutation of that range;
3. decode the resulting rank with factoradic unranking;
4. obtain `n` distinct word indices;
5. map those indices to words.

That is a sound architecture. The factoradic part gives a clean bijection between ranks and ordered selections without replacement. The permutation provides deterministic shuffling, while preserving the one-to-one property.

Important details to settle:

- **Validate `n`**: require `0 ≤ n ≤ 1633`. Decide whether `n=0` should produce an empty name or be rejected.
- **Define position semantics**: positions should normally be zero-based, and positions outside `[0, P(1633,n))` should raise an error or be explicitly wrapped.
- **Handle the permutation domain carefully**: a Feistel network naturally permutes a convenient domain such as `0..2^k-1`, not necessarily the exact namespace size. Cycle walking must repeatedly permute until the result is below `P(1633,n)`.
- **Use a sufficiently wide internal integer representation**: `P(1633,n)` becomes enormous. Avoid fixed-width arithmetic unless you deliberately support only a bounded maximum `n`.
- **Expect cycle-walking cost**: choose the smallest convenient Feistel domain whose size is at least the target range. If the ratio between the domain size and namespace size is close to 1, rejection is cheap; otherwise some inputs may require many iterations.
- **Make the Feistel construction fully specified**: round count, half-width handling, round function, key/seed derivation, integer endianness, and serialization must all be fixed. Otherwise implementations may disagree despite using the same seed.
- **Clarify the role of the seed**: if changing the seed should produce a different ordering, include `seed` in the Feistel key derivation. If names must be unpredictable rather than merely shuffled, use a cryptographically sound keyed permutation; an ordinary hash-based round function may provide reproducibility without security.
- **Ensure exact word-list stability**: the assertion that there are 1,633 words is useful, but the list and ordering are part of the algorithm. Any word-list change changes every generated name.
- **Check duplicate words in the source list**: factoradic unranking guarantees distinct **indices**. It guarantees distinct words only if the wordlist itself contains no duplicate strings.
- **Use arbitrary-precision arithmetic for factoradics**: ranking and unranking require divisions by decreasing radices, and the intermediate values can be much larger than machine integers.
- **Consider output formatting**: decide whether names are space-separated lowercase words and whether the seed is normalized before use.

One subtle point: the “starting position” and “number of successive names” are only an iteration interface. The underlying generator should still accept an individual absolute position, so requesting positions `k` through `k+m-1` produces the same results whether they are requested all at once or separately.

So the central design is correct. The main engineering risks are not the combinatorics; they are specifying the permutation precisely, handling the enormous arbitrary-precision domain, keeping cycle walking efficient, and freezing the word-list/version semantics.

````
import math

def unrank(rank: int, words: list[str], n: int) -> tuple[str, ...]:
    """Convert rank to its corresponding n-word permutation."""
    assert 0 <= n <= len(words)
    namespace_size = math.perm(len(words), n)
    assert 0 <= rank < namespace_size
    
    available = list(range(len(words))) # available word indices
    result = []
    for positions_remaining in range(n, 0, -1):
        # Number of completions for each possible next word
        completions = math.perm(len(available) - 1, positions_remaining - 1)
        choice, rank = divmod(rank, completions)
        result.append(available.pop(choice))
    return tuple(words[i] for i in result)
````

````
import hashlib

def feistel(value: int, bits: int, seed_bytes: bytes) -> int:
    """Permute `value` using a balanced Feistel network:

    1. Split value into equal-width left and right halves.
    2. For each round, hash the seed, round number, and right half.
    3. Mask the hash to the half-width and XOR it into the left half.
    4. Swap the halves.
    5. Recombine them into the permuted value.

    The swaps and XORs make the transformation reversible, hence bijective.
    The result is a bijection of range(2**bits). That is, it rearranges
    the domain without losing or duplicating any values.
    
    `seed_bytes` selects the particular permutation.    
    """
    assert not (bits % 2)
    half_bits = bits // 2
    half_bytes = (half_bits + 7) // 8
    mask = (1 << half_bits) - 1
    left = value >> half_bits
    right = value & mask
    for round_num in range(4):
        message = (
            seed_bytes
            + round_num.to_bytes(1, 'big')
            + right.to_bytes(half_bytes, 'big')
        )
        digest = hashlib.blake2b(message, digest_size=half_bytes).digest()
        round_value = int.from_bytes(digest, 'big') & mask
        left, right = right, left ^ round_value
    return (left << half_bits) | right
````

````
def permute(rank: int, namespace_size: int, seed: str) -> int:
    """Deterministically permute range(namespace_size) by cycle walking.

    Applies a seeded Feistel permutation on the next suitable power-of-two
    domain, then uses cycle walking to return only valid namespace ranks.
    """
    assert 0 <= rank < namespace_size
    bits = (namespace_size - 1).bit_length() # bits needed to represent largest rank
    bits += bits % 2 # ensure event number of bits
    seed_bytes = seed.encode('utf-8')

    while True:
        rank = feistel(rank, bits, seed_bytes)
        if rank < namespace_size:
            return rank
````

````
def generate_name(seed: str, n: int, position: int, words: list[str]=words) -> tuple[str, ...]:
    """Returns a unique `n`-word mnemonic name at `position` in seeded sequence.
    
    Names are deterministic: the same seed + n + position produces the same name.

    For example,
    >>> generate_name('seed', 3, 1_000_000, words)
    ('signal', 'survive', 'flex')
    """
    assert n > 0
    namespace_size = math.perm(len(words), n)
    rank = permute(position, namespace_size, seed)
    return unrank(rank, words, n)
````

Your balanced Feistel domain can be as large as almost four times the valid namespace. That happens because you round the bit width up to an even number. The expected number of Feistel applications is therefore at most about four, which is reasonable, but the tail is unbounded.

For this use case that is probably acceptable. Still, instrumenting the number of iterations would be useful, especially if you later increase `n` or change the domain construction.

BLAKE2b gives a good round function, but four Feistel rounds should not automatically be treated as a cryptographic permutation.
